```python
import tensorflow
from tensorflow import keras
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Flatten
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data()
```

```python
X_train.shape

# in training data, we have already tokenized 20k reviews.
# so if we check it, then we will get data which is already 'integer-decoded'.
```

```python
X_train[0]
```

```python
print(len(X_train[0]))
print(len(X_train[1]))
print(len(X_train[2]))
```

```python
# so first of all, we have to make our all reviews of equal length.
# for that we have to use the 'Padding' 
```

```python
from keras.utils import pad_sequences

X_train = pad_sequences(X_train, padding='post')
X_test = pad_sequences(X_test, padding='post')
```

```python
X_train[0]
```

```python
# creating the model :-

model = Sequential()

# Add SimpleRNN layer with 32 nodes-
model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False))

# Add output layer-
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```
